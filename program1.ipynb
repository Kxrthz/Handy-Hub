```python
name = "Alice"
age = 30
city = "New York"
print(f"Name: {name}")
print(f"Age: {age}")
print(f"City: {city}")
```

```python
num1 = 5
num2 = 10
print(f"Before swap: num1 = {num1}, num2 = {num2}")
temp = num1
num1 = num2
num2 = temp
print(f"After swap: num1 = {num1}, num2 = {num2}")
```

```python
a = 7
b = 14
print(f"Before swap: a = {a}, b = {b}")
a, b = b, a # Pythonic way to swap variables
print(f"After swap: a = {a}, b = {b}")
```

```python
x = y = z = 100
print(f"x = {x}, y = {y}, z = {z}")
```

```python
var1, var2, var3 = "apple", 20, True
print(f"var1 = {var1}, var2 = {var2}, var3 = {var3}")
```

```python
int_var = 10
float_var = 10.5
str_var = "hello"
bool_var = False
print(f"Type of int_var ({int_var}): {type(int_var)}")
print(f"Type of float_var ({float_var}): {type(float_var)}")
print(f"Type of str_var ('{str_var}'): {type(str_var)}")
print(f"Type of bool_var ({bool_var}): {type(bool_var)}")
```

```python
my_variable = "Python"
print(f"Value of my_variable: {my_variable}")
print(f"Memory address of my_variable: {id(my_variable)}")
another_variable = 123
print(f"Value of another_variable: {another_variable}")
print(f"Memory address of another_variable: {id(another_variable)}")
```

```python
global_message = "I am a global variable!"
def display_global():
    print(f"Inside function: {global_message}")
display_global()
print(f"Outside function: {global_message}")
```

```python
def my_function():
    local_variable = "I am local to this function."
    print(f"Inside function: {local_variable}")
my_function()
try:
    print(f"Outside function: {local_variable}")
except NameError as e:
    print(f"Error when accessing local_variable outside function: {e}")
```

```python
my_data = [1, 2, 3]
print(f"Before deletion: my_data = {my_data}")
del my_data
try:
    print(f"After deletion: my_data = {my_data}")
except NameError as e:
    print(f"Error after deleting my_data: {e}")
```
